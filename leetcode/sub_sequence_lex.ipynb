```python
class Solution:
    def smallestSubsequence(self, s: str) -> str:
        # dict to check what shows where
        dict_chars = {}

        # final result stack
        result_stack = []

        # set to track what has been added to the stack
        track_set = set()

        for i in range(len(s)):
            dict_chars[s[i]] = i


        for i in range(len(s)):
            char = s[i]

            if char in track_set:
                continue
            
            # while
            # result_stack is not empty
            # char is lessthan the last char added
            # check if the char will show up again later
            while result_stack and char < result_stack[-1] and dict_chars[result_stack[-1]] > i:
                popped_char = result_stack.pop()
                track_set.remove(popped_char)


            result_stack.append(char)
            track_set.add(char)

        return "".join(result_stack)
    
if __name__ == '__main__':
    sol = Solution()
    # Test your cases here
    print(sol.smallestSubsequence("bcabc"))    # Should print: abc
    print(sol.smallestSubsequence("cbacdcbc")) # Should print: acdb
```

```python
class Solution:
    def smallestSubsequence(self, s: str) -> str:
        dict_chars = {c: i for i, c in enumerate(s)}
        result_stack = []
        track_set = set()
        
        for i, char in enumerate(s):
            if char in track_set:
                continue
            
            # The "Bouncer" logic
            while result_stack and char < result_stack[-1] and dict_chars[result_stack[-1]] > i:
                popped = result_stack.pop()
                track_set.remove(popped)
            
            result_stack.append(char)
            track_set.add(char)
            
        return "".join(result_stack)

if __name__ == '__main__':
    sol = Solution()
    # Test your cases here
    print(sol.smallestSubsequence("bcabc"))    # Should print: abc
    print(sol.smallestSubsequence("cbacdcbc")) # Should print: acdb
```
