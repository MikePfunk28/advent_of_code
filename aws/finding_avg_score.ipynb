```python
if __name__ == '__main__':
    n = int(input())
    student_marks = {}
    for _ in range(n):
        name, *line = input().split()
        scores = list(map(float, line))
        student_marks[name] = scores
    query_name = input()
    #print(query_name)
    #print(student_marks[query_name])
    scores = 0
    for marks in student_marks[query_name]:
        #print(marks)
        scores += marks
        
    #print(scores)
    avg_scores = scores / 3
    print(f"{avg_scores:.2f}")
        
```
