How the Magic Works
args = parts[1:]: This uses list slicing to grab everything after the first word.

For "insert 0 5", args becomes ["5"], index 0, put 5.

For "pop", args becomes an empty list [].

[int(x) for x in args]: This is a list comprehension that loops through whatever strings are in args and turns them into integers.

*int_args (The Unpacking Operator): This is the real star of the show. The * symbol takes a list and unpacks its contents into separate arguments for the function.

If the command is insert, getattr gets my_list.insert. The * unpacks [0, 5] so the code runs my_list.insert(0, 5).

If the command is sort, getattr gets my_list.sort. Since the list is [], it unpacks nothing, running my_list.sort().

This eliminates the need to write an elif for every single list operation!

```python
if __name__ == '__main__':
    N = int(input())
    my_list = []
    for _ in range(N):
        parts = input().split()
        command = parts[0]
        
        if command == "print":
            print(my_list)
        else:
            # 1. Grab everything after the command as arguments
            args = parts[1:]
            
            # 2. Convert all string arguments into integers
            int_args = [int(x) for x in args]
            
            # 3. Use getattr to fetch the method, and * to unpack the arguments
            getattr(my_list, command)(*int_args)
        
```
