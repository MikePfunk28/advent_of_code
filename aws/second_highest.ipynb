## Second highest
First we take in the input and then the list and split it, **.split()**, we take in the entire list at once.  From there we for loop through it, to strip each number.  Then we sort, and map it to int.

Then we strip, **.strip()** it and **convert to set** to *remove duplicates*, then sort and reverse and the first should be the highest and the second index the 2nd highest.  sort() method has a reverse parameter.

```python
if __name__ == '__main__':
    n = int(input())
    # first split the input
    k = input().split()
    
    # Then use strip
    for i in range(len(k)):
        k[i] = k[i].strip()
        #print(k)
    arr = sorted(k)
    arr_nums = map(int, arr)
    #print(arr)
    
    # .sort() returns none so you can write it like
    arr_set = set(arr_nums)
    sorted_nums = sorted(arr_set, reverse=True)
    print(sorted_nums[1])
    
```
