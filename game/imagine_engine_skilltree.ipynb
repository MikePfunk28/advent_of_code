```python
from typing import List
from collections import deque

class Node:
    def __init__(self, data):
        # What do I know at node creation
        # Who Am I?
        self.data = data
        data = []
        # Who am I connected to
        self.neighbors = []


class SkillDag:
    def __init__(self):
        # adjacency list Dictionary
        self.nodes = {}
        
    def addto(self, data):
        self.data = data
        self.nodes[data] = Node(data)
    
    def connectto(self, prerequiste: str, skill: str) -> None:
        prerequiste_node = self.nodes[prerequiste]
        skill_node = self.nodes[skill]
        prerequiste_node.neighbors.append(skill_node)
        
    def get_neighbors(self, data:str)-> List[str]:
        node = self.nodes.get(data)
        if not node:
            return []
        # loops through each neighbor node and pools into a list
        return [n.data for n in node.neighbors]
    
    def bfs(self, start:str) -> List[str]:
        visited = set()
        bfs_q = deque([start])
        result = []
        while bfs_q:
            current = bfs_q.popleft()
            if current in visited:
                continue

            visited.add(current)
            result.append(current)

            for neighbor in self.get_neighbors(current):
                # only queue neighbors that we havent visited
                if neighbor not in visited:
                    bfs_q.append(neighbor)
        
        return result
    
    def dfs(self, start:str) -> List[str]:
        visited = set()
        dfs_stack = [start]
        
        result = []
        while dfs_stack:
            current = dfs_stack.pop()
            if current in visited:
                continue

            visited.add(current)
            result.append(current)

            dfs_stack.extend(reversed(self.get_neighbors))

        return result
    
    def topological_sort(self) -> List[str]:
        # number of incoming edges
        in_degree = {node:0 for node in self.nodes}
        for node in self.nodes:
            for neighbor in self.nodes[node]:
                in_degree[neighbor] += 1
                queue = [node for node in self.nodes if in_degree[node] == 0]
                result = []
        while queue:
            current = queue.pop(0)
            result.append(current)
            for neighbor in self.nodes(current):
                in_degree[neighbor] -= 1
                if in_degree[neighbor] == 0:
                    queue.append(neighbor)
        return result
    
SkillTree = SkillDag()    
```
